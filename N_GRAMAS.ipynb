```python
def generar_ngramas (texto, n):
    texto = texto.lower()

    palabras = texto.split()

    ngramas = []
    print(len(palabras))

    for i in range(len(palabras) - n + 1):
      ngrama = palabras[i:i+n]
      ngramas.append(ngrama)

    return ngramas

texto = "El estudiannte aprende lenguaje natural"

#unigramas
unigramas = generar_ngramas(texto, 1)
bigramas  = generar_ngramas(texto, 2)
trigramas = generar_ngramas(texto, 3)

print('\nUNIGRAMAS:', *unigramas, sep='\n')
print('\nBIGRAMAS:', *bigramas, sep='\n')
print('\nTRIIGRAMAS:',*trigramas, sep='\n')
```

```python
from nltk import ngrams

texto = input("Escriba una oracion: ")
n = int(input("Indica el valor de n: "))

tokens = texto.lower().split()
resultado = list(ngrams(tokens, n))

print(f"\n{n}-gramas encontrados: ")
for ng in resultado:
  print(ng)
```

```python
from nltk import ngrams

texto = input("Escriba una oracion: ")

tokens = texto.lower().split()
for n in range(1, len(tokens) + 1):
  resultado = list(ngrams(tokens, n))

  print(f"\n{n}-gramas encontrados: ")
  for ng in resultado:
    print(ng)
```

#CONCLUSIONES

## 1. **¿Qué representa la letra N en un n-grama?**
La letra N representa el número de elementos consecutivos que se agrupan en una secuencia o ventana deslizante. Por ejemplo, si N=2, analizamos pares de palabras consecutivas.

## 2. **¿Cuál es la diferencia entre un unigrama, un bigrama y un trigrama?**
La diferencia es el tamaño de las secuencias
* **Unigrama (N=1):** Palabras individuales de forma aislada (ej. `['el']`, `['lenguaje']`). No capturan contexto de adyacencia.
* **Bigrama (N=2):** Parejas de palabras consecutivas (ej. `['el', 'estudiante']`).
* **Trigrama (N=3):** Tríos de palabras consecutivas (ej. `['el', 'estudiante', 'aprende']`).

## 3. **¿Qué función de NLTK permite generar n-gramas?**
La función `ngrams` del módulo `nltk` (`from nltk import ngrams`). Esta toma como parámetros una secuencia de tokens y el valor entero de `n`.

## 4. **¿Por qué un trigrama conserva más contexto que un bigrama?**
Porque un trigrama considera la relación y el orden de tres palabras consecutivas en lugar de solo dos. Al tener una ventana de contexto más amplia, captura mejor las dependencias sintácticas y semánticas locales de la oración (por ejemplo, el sujeto, el verbo y el objeto juntos).

## 5. **¿Qué significa P(estudia | alumno)?**
Representa la **probabilidad condicional** de que aparezca la palabra *"estudia"* dado que la palabra anterior fue *"alumno"*. En un modelo de lenguaje de bigramas, se calcula como:
$$P(\text{estudia} \mid \text{alumno}) = \frac{\text{Frecuencia}(\text{alumno estudia})}{\text{Frecuencia}(\text{alumno})}$$

## 6. **¿Qué problema puede aparecer cuando N aumenta demasiado?**
* **Dispersión de datos (Sparsity):** A medida que N crece, la cantidad de n-gramas posibles aumenta exponencialmente, lo que hace que la mayoría de las combinaciones tengan frecuencia cero en nuestro corpus de entrenamiento (nunca antes vistas), provocando probabilidades de cero.
* **Consumo de recursos:** Se requiere mucha más memoria y capacidad de cómputo para almacenar y procesar las tablas de frecuencias de n-gramas largos.
* **Pérdida de generalización:** El modelo se vuelve tan específico al texto de entrenamiento que pierde la capacidad de predecir o evaluar textos nuevos de manera flexible (sobreajuste).

## 7. **Modifica el ejemplo de Colab para generar 4-gramas y compara los resultados.**


```python
from nltk import ngrams

texto = input("Escriba una oracion: ")
n = 4

tokens = texto.lower().split()
resultado = list(ngrams(tokens, n))

print(f"\n{n}-gramas encontrados: ")
for ng in resultado:
  print(ng)
```

**Comparación:**
* Con **N=1, 2 o 3**, obtenemos más cantidad de n-gramas (5, 4 y 3 respectivamente) pero con menor contexto por grupo.
* Con **N=4**, solo obtenemos 2 grupos. Aunque mantienen un contexto semántico muy rico de la estructura de la oración original, son secuencias tan específicas que es mucho menos probable encontrarlas idénticas en otro texto.
