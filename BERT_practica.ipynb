## 1. Clasificación de sentimientos

```python
from transformers import pipeline

clasificador = pipeline("sentiment-analysis")

texto = "La pelicula fue excelente"
resultado = clasificador(texto)
print(resultado)
```

## 2. Reconocimiento de entidades (NER)

```python
from transformers import pipeline

ner = pipeline(
    "ner",
    model="dbmdz/bert-large-cased-finetuned-conll03-english"
)
texto = "María visitó Google en México."
resultado = ner(texto)
print(resultado)
```

## 3. Preguntas y respuestas

```python
from transformers import pipeline

# Usamos text-generation con un modelo de lenguaje para responder la pregunta
generador = pipeline("text-generation", model="TinyLlama/TinyLlama-1.1B-Chat-v1.0")

contexto = "Ciudad Guzmán es una ciudad en Jalisco, México."
pregunta = "¿En qué estado está Ciudad Guzmán?"

prompt = f"Contexto: {contexto}\n\nPregunta: {pregunta}\n\nRespuesta corta:"
resultado = generador(prompt, max_new_tokens=20, temperature=0.1, do_sample=False)

print(resultado[0]['generated_text'])
```
