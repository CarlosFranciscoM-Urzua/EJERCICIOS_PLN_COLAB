```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer

#CORPUS
corpus = [
    'el coche es rapido',
    'el coche es azul',
    'un coche rapido azul'
]

vectorizador = CountVectorizer()
matriz_dispersa = vectorizador.fit_transform(corpus)

df_unigramas = pd.DataFrame(
    matriz_dispersa.toarray(),
    columns=vectorizador.get_feature_names_out()
)
df_unigramas.index = ['Documento 1', 'Documento 2', 'Documento 3']

print("Matriz de Unigramas")
display(df_unigramas)

#BIGRAMAS
vectorizador_bigramas = CountVectorizer(ngram_range=(2,2))
matriz_bigramas = vectorizador_bigramas.fit_transform(corpus)

df_bigramas = pd.DataFrame(
    matriz_bigramas.toarray(),
    columns = vectorizador_bigramas.get_feature_names_out()
)
df_bigramas.index = ['Documento 1', 'Documento 2', 'Documento 3']

print('MATRIZ DE BIGRAMAS')
display(df_bigramas)

#FUNCION PARA CALCULAR SPARCITY
def calcular_sparcity(matriz):
    total_celdas = matriz.size
    cantidad_ceros = np.count_nonzero(matriz == 0)
    return cantidad_ceros / total_celdas

sparsity_unigramas = calcular_sparcity(df_unigramas.to_numpy())
sparsity_bigramas = calcular_sparcity(df_bigramas.to_numpy())

#RESUMEN
resumen = pd.DataFrame({
    'Representacion': ['Unigramas', 'Bigramas'],
    'Documentos': [df_unigramas.shape[0], df_bigramas.shape[0]],
    'Caracteristicas': [df_unigramas.shape[1], df_bigramas.shape[1]],
    'Sparcity': [sparsity_unigramas, sparsity_bigramas]
})

resumen['Sparcity'] = resumen['Sparcity'].map(lambda x: f"{x:.2%}")

print('COMPARACION')
display(resumen)
```
